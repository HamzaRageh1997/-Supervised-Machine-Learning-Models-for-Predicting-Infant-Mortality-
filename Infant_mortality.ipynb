```python
dataMapping_Mortality={
    "Dead" : 1,
    "Alive" : 0
}
```

```python
Infant['Mortality'] = Infant['Child alive or dead at time of interview'].map(dataMapping_Mortality)
Infant
```

```python
Infant = Infant[['Educationlevel', 'Number of household members (listed)', 'Sex of household head', 'Total children ever born',  'Age of respondent at 1st birth','Size of child', 'Birth Order', 'Gender of Child',
       'Current age in months', 'Mortality']]
```

```python
Infant
```

```python
x_Infant = Infant[['Educationlevel', 'Number of household members (listed)', 'Sex of household head', 'Total children ever born', 'Size of child', 'Age of respondent at 1st birth', 'Birth Order', 'Gender of Child',
      'Current age in months']]

y_Infant = Infant[['Mortality']]
```

```python
from imblearn.over_sampling import RandomOverSampler
import matplotlib.pyplot as plt

ros = RandomOverSampler(sampling_strategy='not majority')
x_Infant, y_Infant = ros.fit_resample(x_Infant, y_Infant)

ax = y_Infant.value_counts().plot.pie(autopct = '%.2f')
_ = ax.set_title("Infant Mortality")
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_Infant, y_Infant,test_size=0.30, random_state=15)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.naive_bayes import GaussianNB

# Logistic Regression
logreg = LogisticRegression()
logreg.fit(x_train, y_train)

# Decision Tree
dt = DecisionTreeClassifier()
dt.fit(x_train, y_train)

# Random Forests
rf = RandomForestClassifier()
rf.fit(x_train, y_train)

# SVM
svm = SVC()
svm.fit(x_train, y_train)

# Neural Network
nn = MLPClassifier()
nn.fit(x_train, y_train)

# Naive Bayes
nb = GaussianNB()
nb.fit(x_train, y_train)
```
